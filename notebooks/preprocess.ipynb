My text contains abbreviations like Mr., Mrs. and etc. During tokenization, this was inconvenient, so I will replace these abbreviations with the full words. This issue only occurs in two out of the three texts: eng_Anne_of_green_gables.txt and spa_Anne_of_green_gables.txt. The text rus_Anne_of_green_gables.txt does not have this problem. The texts contained chapter titles and standalone lines like 'Chap X', as well as license information, names of typesetters and/or translators, annotations, cover image captions, and tables of contents. I manually removed all of these parts from the texts.

# 2. spanish version

```python
input_path = r'C:\Users\Ксения\Project\Course_NLP\data\without_chap_and_title\spa_Anne_of_green_gables.txt'
output_path = r'C:\Users\Ксения\Project\Course_NLP\data\without_chap_and_title\1spa_Anne_full_abbr.txt'
```

```python
# abberviantion and spanish analogues
abbreviation_map = {
    'Mr.': 'Señor',
    'Ms.': 'Señorita',  # 'Señora' или 'Sra.'
    'Mrs.': 'Señora',
    'Dr.': 'Doctor',
    'St.': 'San',  # for 'saint' - 'San Juan'
    'Sr.': 'Sr.',  
    'Prof.': 'Profesor',
    'Gen.': 'General',
    'Rep.': 'Representante',
    'Sen.': 'Senador',
    'Gov.': 'Gobernador',
    'Lt.': 'Teniente',
    'Col.': 'Coronel',
    'Maj.': 'Mayor',
    'Capt.': 'Capitán',
}
```

```python
# read spa_Anne
with open(input_path, 'r', encoding='utf-8') as f:
    text = f.read()

# Замена сокращений
for abbr, full in abbreviation_map.items():
    text = text.replace(abbr, full)

# save new version
with open(output_path, 'w', encoding='utf-8') as f:
    f.write(text)
```
